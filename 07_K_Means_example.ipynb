# ✨ The K-Means Clustering Script for English Digits ✨

The script provides a complete implementation of K-Means on the digits dataset (handwritten English digits 0–9).

## 🧠 Purpose of the Script

Unsupervised learning to discover the natural structure of digits 0–9 without using real labels.


We want to see if the algorithm can create clusters that correspond to actual digits.

## 🔧 Steps Performed in the Script

### Preprocess

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
X = scaler.fit_transform(X)
```

```python
X
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, label, test_size=0.3, random_state=42, shuffle=True)
```

### Run K-Means


#### K-Means++

```python
from sklearn.cluster import KMeans
```

```python
# Create an instance of a model
K_means = KMeans(n_clusters=10, init="k-means++", n_init="auto")
```

```python
K_means.fit(X_train)
```

```python
train_clusters = K_means.labels_
```

```python
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA


pca = PCA(n_components=2, random_state=42)
X_train_pca = pca.fit_transform(X_train)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

scatter1 = ax1.scatter(X_train_pca[:, 0], X_train_pca[:, 1], 
                        c=train_clusters, cmap='tab10', alpha=0.6, s=20)
ax1.set_title("K-means clustering on training data")
ax1.set_xlabel("PC1")
ax1.set_ylabel("PC2")
plt.colorbar(scatter1, ax=ax1, ticks=range(10), label="Cluster ID")

scatter2 = ax2.scatter(X_train_pca[:, 0], X_train_pca[:, 1], 
                        c=y_train, cmap='tab10', alpha=0.6, s=20)
ax2.set_title("Real Numbers (Field of Truth)")
ax2.set_xlabel("PC1")
ax2.set_ylabel("PC2")
plt.colorbar(scatter2, ax=ax2, ticks=range(10), label="True Digit")

plt.tight_layout()
plt.show()
```

```python
y_pred = K_means.predict(X_test)
```

#### Evaluation for K-Means++

```python
from sklearn.metrics import adjusted_rand_score, adjusted_mutual_info_score, homogeneity_score
from sklearn.metrics import silhouette_score, completeness_score, v_measure_score
```

```python
ari = adjusted_rand_score(y_test, y_pred)
ami = adjusted_mutual_info_score(y_test, y_pred)
hom = homogeneity_score(y_test, y_pred)
com = completeness_score(y_test, y_pred)
vm = v_measure_score(y_test, y_pred)
sil = silhouette_score(X_test, y_test)
```

```python
print(f"{ari:.3f}")
print(f"{ami:.3f}")
print(f"{hom:.3f}")
print(f"{com:.3f}")
print(f"{vm:.3f}")
print(f"{sil:.3f}")
```
